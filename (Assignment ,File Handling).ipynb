### Q14 
Ek program banayein jisme `finally` block ho, jo humesha execute ho chahe error aaye ya na aaye.

```python
try:
    num1 = int(input("Enter a number"))
    num2 = int(input("Enter a number"))
    result = num1 /num2
except ValueError:
    print("please Enter an Right Value")
except ZeroDivisionError:
    print("The Number is not divided by zero")
finally:
    print("Program Finished")
```
